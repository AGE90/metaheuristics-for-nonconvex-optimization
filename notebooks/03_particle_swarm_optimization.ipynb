# Particle Swarm Optimization

Kennedy & Eberhart (1995), with the Shi & Eberhart (1998) linearly-decaying inertia weight:

$$v_i \leftarrow w v_i + c_1 r_1 (p_i - x_i) + c_2 r_2 (g - x_i), \qquad x_i \leftarrow x_i + v_i$$

Each particle is pulled towards its own best-known position $p_i$ and the swarm's best $g$. See `docs/notes/particle-swarm-optimization.md` for details.

```python
%load_ext autoreload
%autoreload 2
```

```python
import numpy as np
import plotly.graph_objects as go

from metaheuristics.algorithms.particle_swarm import ParticleSwarmOptimization
from metaheuristics.benchmarks import landscapes as L
from metaheuristics.viz.plotly_surfaces import plot_contour_with_trajectory
```

## Ackley

PSO tends to do very well here — the swarm quickly funnels into the central well.

```python
np.random.seed(42)
optimizer = ParticleSwarmOptimization(num_particles=40, max_iterations=100)
result = optimizer.optimize(L.ackley, L.ackley.bounds)
print('best solution:', result.best_solution)
print('best fitness :', result.best_fitness)
```

```python
go.Figure(go.Scatter(y=result.fitness_history)).update_layout(
    title='Particle Swarm Optimization convergence on Ackley', xaxis_title='iteration', yaxis_title='best fitness'
).show()
```

```python
plot_contour_with_trajectory(L.ackley, result).show()
```

## Eggholder

A much harder landscape: watch how the swarm's global best can get stuck in a strong local basin.

```python
np.random.seed(42)
optimizer = ParticleSwarmOptimization(num_particles=40, max_iterations=100)
result = optimizer.optimize(L.eggholder, L.eggholder.bounds)
print('best solution:', result.best_solution)
print('best fitness :', result.best_fitness)
```

```python
go.Figure(go.Scatter(y=result.fitness_history)).update_layout(
    title='Particle Swarm Optimization convergence on Eggholder', xaxis_title='iteration', yaxis_title='best fitness'
).show()
```

```python
plot_contour_with_trajectory(L.eggholder, result).show()
```
