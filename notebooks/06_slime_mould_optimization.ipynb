# Slime Mould Algorithm

Li, Chen, Deng & Mirjalili (2020), inspired by *Physarum polycephalum* foraging. Agents either approach the best food source, weighted by a 'smell' term, or contract/oscillate around their own position:

$$X(t+1) = \begin{cases} X_b(t) + \vec{vb}\cdot(W\cdot X_A(t) - X_B(t)) & r < p \\ \vec{vc}\cdot X(t) & r \geq p \end{cases}$$

See `docs/notes/slime-mould-optimization.md` for what each term means.

```python
%load_ext autoreload
%autoreload 2
```

```python
import numpy as np
import plotly.graph_objects as go

from metaheuristics.algorithms.slime_mould import SlimeMouldAlgorithm
from metaheuristics.benchmarks import landscapes as L
from metaheuristics.viz.plotly_surfaces import plot_contour_with_trajectory
```

## Griewank

```python
np.random.seed(42)
optimizer = SlimeMouldAlgorithm(population_size=40, max_iterations=100)
result = optimizer.optimize(L.griewank, L.griewank.bounds)
print('best solution:', result.best_solution)
print('best fitness :', result.best_fitness)
```

```python
go.Figure(go.Scatter(y=result.fitness_history)).update_layout(
    title='Slime Mould Algorithm convergence on Griewank', xaxis_title='iteration', yaxis_title='best fitness'
).show()
```

```python
plot_contour_with_trajectory(L.griewank, result).show()
```
