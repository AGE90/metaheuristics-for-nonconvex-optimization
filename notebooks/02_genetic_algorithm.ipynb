# Genetic Algorithm

A binary-encoded GA (Holland, 1975): each decision variable is encoded as a fixed-length bitstring and linearly decoded back into its bounded range,

$$x = lb + \frac{\text{int}(b)}{2^{L}-1}(ub - lb)$$

so the same bit-flip mutation / single-point crossover / tournament-selection machinery that works for combinatorial problems (e.g. OneMax) can also minimize continuous benchmark functions. See `docs/notes/genetic-algorithm.md` for the full write-up and `src/metaheuristics/algorithms/genetic_algorithm.py` for the implementation.

```python
%load_ext autoreload
%autoreload 2
```

```python
import numpy as np
import plotly.graph_objects as go

from metaheuristics.algorithms.genetic_algorithm import GeneticAlgorithm
from metaheuristics.benchmarks import landscapes as L
from metaheuristics.viz.plotly_surfaces import plot_contour_with_trajectory
```

## Minimizing Rastrigin

```python
np.random.seed(42)
ga = GeneticAlgorithm(population_size=60, max_generations=150, mutation_rate=0.02, bits_per_variable=20)
result = ga.optimize(L.rastrigin, L.rastrigin.bounds)
print('best solution:', result.best_solution)
print('best fitness :', result.best_fitness)
```

```python
go.Figure(go.Scatter(y=result.fitness_history)).update_layout(
    title='GA convergence on Rastrigin', xaxis_title='generation', yaxis_title='best fitness'
).show()
```

```python
plot_contour_with_trajectory(L.rastrigin, result).show()
```

## A harder case: Eggholder

Eggholder's extreme ruggedness is a good stress test for how much elitism and population size matter.

```python
np.random.seed(42)
ga = GeneticAlgorithm(population_size=100, max_generations=300, mutation_rate=0.03, bits_per_variable=20)
result = ga.optimize(L.eggholder, L.eggholder.bounds)
print('best solution:', result.best_solution)
print('best fitness :', result.best_fitness, ' (global minimum:', L.eggholder.global_minimum[1], ')')
```

```python
plot_contour_with_trajectory(L.eggholder, result).show()
```
