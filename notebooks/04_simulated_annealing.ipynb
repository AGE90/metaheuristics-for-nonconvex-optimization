# Simulated Annealing

Kirkpatrick, Gelatt & Vecchi (1983). A single point wanders the landscape, proposing Gaussian-perturbed candidates and accepting worse ones with probability

$$P(\text{accept}) = \exp\left(-\frac{f(x') - f(x)}{T}\right), \qquad T \leftarrow \alpha T$$

Unlike GA/PSO/DE/SMO, this is **not population-based** — expect noisier, less reliable convergence on multimodal landscapes. See `docs/notes/simulated-annealing.md`.

## Rastrigin

Compare this convergence curve's noisiness to the population-based algorithms in the other notebooks.

```python
np.random.seed(42)
optimizer = SimulatedAnnealing(max_iterations=3000, initial_temperature=10.0, cooling_rate=0.995)
result = optimizer.optimize(L.rastrigin, L.rastrigin.bounds)
print('best solution:', result.best_solution)
print('best fitness :', result.best_fitness)
```

```python
go.Figure(go.Scatter(y=result.fitness_history)).update_layout(
    title='Simulated Annealing convergence on Rastrigin', xaxis_title='iteration', yaxis_title='best fitness'
).show()
```

```python
plot_contour_with_trajectory(L.rastrigin, result).show()
```
