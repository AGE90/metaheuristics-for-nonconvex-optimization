# Differential Evolution

Storn & Price (1997), DE/rand/1/bin. Mutants are built from the *difference* of two random population members:

$$v = a + F (b - c)$$

then combined with the target vector via binomial crossover. DE is known for being particularly effective on valley-shaped landscapes like Rosenbrock. See `docs/notes/differential-evolution.md`.

```python
import numpy as np
import plotly.graph_objects as go

from metaheuristics.algorithms.differential_evolution import DifferentialEvolution
from metaheuristics.benchmarks import landscapes as L
from metaheuristics.viz.plotly_surfaces import plot_contour_with_trajectory
```

## Rosenbrock

DE's difference-vector mutation is well suited to following curved valleys like this one.

```python
np.random.seed(42)
optimizer = DifferentialEvolution(population_size=40, max_generations=150)
result = optimizer.optimize(L.rosenbrock, L.rosenbrock.bounds)
print('best solution:', result.best_solution)
print('best fitness :', result.best_fitness)
```

```python
go.Figure(go.Scatter(y=result.fitness_history)).update_layout(
    title='Differential Evolution convergence on Rosenbrock', xaxis_title='iteration', yaxis_title='best fitness'
).show()
```

```python
plot_contour_with_trajectory(L.rosenbrock, result).show()
```
