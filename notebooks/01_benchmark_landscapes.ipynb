# Benchmark Landscapes

Interactive 3D surfaces of the nonconvex test functions used throughout this project (rotate/zoom each plot). All functions live in `metaheuristics.benchmarks.landscapes`, are posed as **minimization** problems, and carry `.bounds` / `.global_minimum` metadata used by the optimizer notebooks and the test suite. See `docs/notes/benchmark-functions.md` for the full reference table.

```python
from metaheuristics.benchmarks import landscapes as L
from metaheuristics.viz.plotly_surfaces import plot_surface
```

## Sphere

$$f(\mathbf{x}) = \sum_i x_i^2$$

Bowl-shaped, unimodal. The baseline sanity check: every algorithm should nail this.

```python
plot_surface(L.sphere).show()
```

## Rastrigin

$$f(\mathbf{x}) = An + \sum_i \left[x_i^2 - A\cos(2\pi x_i)\right], \quad A=10$$

Many local minima arranged on a regular grid over a parabolic envelope.

```python
plot_surface(L.rastrigin).show()
```

## Ackley

$$f(x,y) = -20e^{-0.2\sqrt{0.5(x^2+y^2)}} - e^{0.5(\cos 2\pi x + \cos 2\pi y)} + e + 20$$

Many local minima, nearly flat outer region with a sharp central well.

```python
plot_surface(L.ackley).show()
```

## Rosenbrock

$$f(x,y) = (a-x)^2 + b(y-x^2)^2, \quad a=1,\ b=100$$

The 'banana' valley: easy to find the valley, hard to converge to its minimum.

```python
plot_surface(L.rosenbrock).show()
```

## Beale

$$f(x,y) = (1.5-x+xy)^2 + (2.25-x+xy^2)^2 + (2.625-x+xy^3)^2$$

Multimodal with a narrow curved valley near the optimum.

```python
plot_surface(L.beale).show()
```

## Griewank

$$f(\mathbf{x}) = \sum_i \frac{x_i^2}{4000} - \prod_i \cos\left(\frac{x_i}{\sqrt{i+1}}\right) + 1$$

Many local minima; the product term makes dimensions non-separable.

```python
plot_surface(L.griewank).show()
```

## Himmelblau

$$f(x,y) = (x^2+y-11)^2 + (x+y^2-7)^2$$

Four global minima of equal value — good for checking if a run diversifies or collapses to one basin.

```python
plot_surface(L.himmelblau).show()
```

## Easom

$$f(x,y) = -\cos(x)\cos(y)\exp\left(-((x-\pi)^2+(y-\pi)^2)\right)$$

A huge flat plateau with a single steep needle-shaped optimum — stresses exploration.

```python
plot_surface(L.easom).show()
```

## Eggholder

$$f(x,y) = -(y+47)\sin\sqrt{|x/2+(y+47)|} - x\sin\sqrt{|x-(y+47)|}$$

Extremely rugged; one of the hardest common 2D benchmarks.

```python
plot_surface(L.eggholder).show()
```

## Booth

$$f(x,y) = (x+2y-7)^2 + (2x+y-5)^2$$

Simple convex-looking bowl, mainly used as a quick sanity check.

```python
plot_surface(L.booth).show()
```

## Matyas

$$f(x,y) = 0.26(x^2+y^2) - 0.48xy$$

Plate-shaped and smooth — nearly convex.

```python
plot_surface(L.matyas).show()
```

## Levi13

$$f(x,y) = \sin^2(3\pi x) + (x-1)^2(1+\sin^2(3\pi y)) + (y-1)^2(1+\sin^2(2\pi y))$$

Many local minima layered on top of a smoother global trend.

```python
plot_surface(L.levi13).show()
```
